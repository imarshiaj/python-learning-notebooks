## 1. What is Python?

Python is a general-purpose programming language known for readable syntax and a large ecosystem. People use it for automation, data analysis, web applications, scientific computing, artificial intelligence, and many other tasks.

A program is a set of instructions that a computer follows. Python lets us express those instructions in a form that is approachable for humans.

## 2. Where can you write Python?

You can use a local editor such as VS Code, an interactive Python shell, or a notebook environment such as Jupyter. A notebook combines explanatory text, code, and results in one document.

For local notebooks, install Python and then install Jupyter with `python -m pip install notebook`, or use an existing hosted notebook environment. This project does not require third-party Python libraries for its examples.

## 3. Your first program

The built-in `print()` function displays information. Text must be placed inside quotes.

```python
print("Hello, Python!")
print("Welcome to programming.")
print(2 + 3)
```

The first two lines display text. The third line evaluates a mathematical expression and displays its result. Parentheses contain the arguments passed to `print()`.

## 4. Comments and readable code

A comment begins with `#`. Python ignores the comment when executing the program. Comments should explain intent or context rather than repeat obvious code.

```python
# Calculate the number of minutes in two hours.
minutes_per_hour = 60
hours = 2
print(minutes_per_hour * hours)
```

## 5. Expressions, values, and types

An **expression** is code that produces a value. Common built-in types include `int` (integer), `float` (floating-point number), `str` (text), and `bool` (`True` or `False`). The `type()` function helps you inspect a value.

```python
print(10 + 5)
print(type(10))
print(type(3.14))
print(type("Python"))
print(type(True))
```

## 6. A first look at variables

A variable is a name that refers to a value. You will study naming and assignment in detail in Notebook 03. For now, notice how a descriptive name can make code easier to read.

```python
language = "Python"
print("I am learning", language)
```

## 7. Common beginner mistakes

- Text needs matching quotation marks: `"Hello"` or `'Hello'`.
- Python is case-sensitive: `name` and `Name` are different names.
- Indentation is meaningful in Python. Later, conditionals, loops, and functions will use indentation to define blocks.
- A `SyntaxError` usually means the code does not follow Python's grammar. Read the line and location mentioned in the message.

## 9. More useful everyday examples

A few small built-in functions can answer common questions without requiring a library. `len()` counts items, `round()` rounds a number for display or a calculation, and `abs()` returns a non-negative distance from zero.

```python
label = "Python basics"
values = [12, 18, 15]

print("Label:", label)
print("Characters:", len(label))
print("Average:", round(sum(values) / len(values), 2))
print("Distance from zero:", abs(-17))
```

### Separate output for readability

`print()` accepts multiple arguments. The `sep` parameter controls the separator, while `end` controls what is printed at the end. The defaults are a space and a newline.

```python
print("2026", "10", "19", sep="-")
print("Loading", end="... ")
print("done")
```

### Recognize common beginner errors

- `SyntaxError`: Python cannot parse the code, often because a quote, colon, or parenthesis is missing.
- `NameError`: a name has not been defined or was typed with the wrong capitalization.
- `TypeError`: an operation does not support the supplied types, such as adding a string to an integer.

This example catches two errors intentionally so you can see their names without stopping the notebook. In real code, catch only errors you can handle meaningfully.

```python
examples = [
    ("undefined name", lambda: name_that_does_not_exist),
    ("incompatible types", lambda: "age: " + 21),
]

for description, operation in examples:
    try:
        operation()
    except (NameError, TypeError) as error:
        print(description, "->", type(error).__name__)
```
