## 1. Make decisions with `if`

Conditional statements run a block only when a condition is true. Python uses a colon after the condition and indentation to define the block. Four spaces per indentation level is the usual convention.

```python
temperature = 18
if temperature > 20:
    print("It is warm.")
else:
    print("It is not above 20 degrees.")
```

## 2. Use `if`, `elif`, and `else`

Use `elif` for additional alternatives. Python checks the conditions from top to bottom and executes the first matching branch. `else` is optional and handles the case where no preceding condition is true.

```python
score = 84
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "Needs more practice"
print(grade)
```

## 3. Combine conditions

Use comparison operators with `and`, `or`, and `not`. Put parentheses around complex conditions when that improves readability. Be careful to express the actual logic you intend.

```python
age = 24
has_ticket = True
if age >= 18 and has_ticket:
    print("Entry permitted.")

wind_speed = 8
rain_expected = False
if wind_speed > 20 or rain_expected:
    print("Review the outdoor plan.")
else:
    print("No weather warning from these checks.")
```

## 4. Truthy and falsy values

Python treats some values as false in conditions: `False`, `None`, numeric zero, and empty collections such as `""`, `[]`, `{}`, and `set()`. Most other objects are truthy. You can often check whether a collection is non-empty with `if items:` rather than `if len(items) > 0:`.

```python
items = []
if items:
    print("There are items.")
else:
    print("The collection is empty.")

user_name = "Morgan"
if user_name:
    print(f"Hello, {user_name}!")
```

## 5. Conditional expressions

A conditional expression selects one of two values: `value_if_true if condition else value_if_false`. Use it for simple decisions, not complicated nested logic.

```python
score = 72
status = "pass" if score >= 60 else "retry"
print(status)
```

## 6. Check several possible values

The `in` operator can make conditions more readable when checking whether a value belongs to a collection of allowed values.

```python
status = "in_progress"
if status in {"queued", "in_progress", "complete"}:
    print("Known status")
else:
    print("Unknown status")
```

## Common mistakes

- Use `==` to compare values, not `=`.
- Indentation defines code blocks; inconsistent indentation can cause errors or change program logic.
- The order of `if`/`elif` branches matters. Put more specific conditions before broader ones when necessary.
- Avoid conditions that are difficult to parse. Use intermediate variables to give important checks meaningful names.

## 7. Check `None` explicitly

Use `is None` or `is not None` to test whether a value is the special object `None`. Do not use `== None` as a general identity check. Also, do not use `if value` when zero or an empty string is a valid value that must be distinguished from missing data.

```python
distance = 0
optional_label = None

if distance is not None:
    print("Distance was provided:", distance)

if optional_label is None:
    print("No label was supplied.")
```

## 8. Short-circuit evaluation

With `and`, Python stops as soon as a condition is false. With `or`, it stops as soon as a condition is true. Short-circuiting can prevent an operation from running when its prerequisite is not met.

```python
denominator = 0
if denominator != 0 and 10 / denominator > 1:
    print("The ratio is greater than one.")
else:
    print("The ratio is undefined or not greater than one.")

name = ""
print(name or "Unknown name")
```

## 9. Condition order matters

An `if`/`elif` chain uses the first true branch. Put more specific or higher-priority conditions before broad ones; otherwise a broad condition can make later branches unreachable.

```python
score = 95
if score >= 60:
    result = "Pass"
elif score >= 90:
    result = "Excellent"  # Never reached for score 95.
else:
    result = "Fail"
print(result)

# Correct order: check the higher threshold first.
if score >= 90:
    result = "Excellent"
elif score >= 60:
    result = "Pass"
else:
    result = "Fail"
print(result)
```
