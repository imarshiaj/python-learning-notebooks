## 1. What is a module?

A module is a Python file that contains code such as functions, classes, and constants. Importing modules helps organize code and reuse functionality. Python includes a large standard library, so many tasks do not require third-party packages.

````python
import math
import statistics as stats
from datetime import date

measurements = [10, 12, 12, 14, 17]
print(math.sqrt(81))
print(stats.mean(measurements))
print(date.today().isoformat())
````

## 2. Import styles

Common forms include `import module`, `import module as alias`, and `from module import name`. Prefer explicit imports that make the source of a name clear. Avoid `from module import *` in most application code because it can make names ambiguous.

````python
import random as rng
from pathlib import Path

print(rng.choice(["north", "south", "east", "west"]))
print(Path("data") / "input.csv")
````

## 3. The `__name__` convention

When Python runs a file directly, its special variable `__name__` is set to `"__main__"`. When the file is imported as a module, `__name__` is normally the module's import name. This pattern lets a file provide reusable definitions without running demonstration code on every import.

````python
def main():
    print("Run the program's main task here.")

if __name__ == "__main__":
    main()
````

## 4. Create and import a small custom module

A module is simply a `.py` file. This example creates a temporary module, imports it by its file path, calls one of its functions, and then removes the temporary directory automatically. It demonstrates the idea without adding generated files to your project.

````python
from pathlib import Path
from tempfile import TemporaryDirectory
import importlib.util

with TemporaryDirectory() as temp_dir:
    module_path = Path(temp_dir) / "geometry.py"
    module_path.write_text(
        "def rectangle_area(width, height):\n    return width * height\n",
        encoding="utf-8",
    )
    spec = importlib.util.spec_from_file_location("geometry", module_path)
    geometry = importlib.util.module_from_spec(spec)
    spec.loader.exec_module(geometry)
    print(geometry.rectangle_area(4, 3))
````

## 5. What is a package?

A package organizes related modules in a directory. A simple traditional package often contains an `__init__.py` file. Python also supports namespace packages, but `__init__.py` remains a useful explicit convention for beginner projects. For example:

```text
my_project/
├── README.md
├── main.py
└── tools/
    ├── __init__.py
    ├── geometry.py
    └── text_utils.py
```

Code in `main.py` could import a function with `from tools.geometry import rectangle_area` when the project is run with the project root on the import path.

## 6. Install third-party packages

Python's package installer is `pip`. In a project, use a virtual environment to isolate dependencies. A common workflow is:

```bash
python -m venv .venv
# Activate the environment using the instructions for your operating system.
python -m pip install package_name
python -m pip freeze > requirements.txt
```

Do not install every package globally by default. For reproducible projects, document the Python version and the dependencies actually needed.

## 7. Import paths and naming

Python searches locations listed in `sys.path` when resolving imports. The current working directory often matters in notebooks and scripts. Avoid naming your own files after standard-library modules such as `random.py`, `statistics.py`, or `json.py`, because the local file can shadow the intended module.

````python
import sys
print("Python version:", sys.version.split()[0])
print("Number of import locations:", len(sys.path))
````

## 7. Avoid naming conflicts

Do not name your own file after a standard-library module you intend to import, such as `random.py`, `json.py`, or `statistics.py`. Python may import your file instead of the standard module. If an import behaves strangely, check the current directory and module filename.

````python
import math as math_tools
from pathlib import Path

print(math_tools.ceil(3.2))
print(Path.cwd().name or "Root directory")
````

## 8. The `__main__` guard

A file's `if __name__ == "__main__":` block runs when the file is executed directly, but not when it is imported as a module. This lets a module provide reusable functions while keeping demonstration code out of normal imports.

````python
def describe_language():
    return "Python supports modules and packages."

if __name__ == "__main__":
    print(describe_language())
````

## 9. Install packages into the intended environment

Use `python -m pip install package_name` so `pip` is associated with the Python interpreter named by `python`. A virtual environment helps keep project dependencies separate. Do not install a package simply because an import fails until you check whether it is part of the standard library or whether the active interpreter is the expected one.

````python
import sys
print("Interpreter:", sys.executable)
print("Python version:", sys.version.split()[0])
````

### Typical import errors

- `ModuleNotFoundError`: the module cannot be found in the active environment or import path.
- `ImportError`: a module was found, but the requested name may not be available.
- Circular imports can occur when two modules depend on each other during initialization. Move shared code into a third module or restructure the dependencies.
