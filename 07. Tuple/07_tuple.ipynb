## 1. What is a tuple?

A tuple is an ordered, immutable collection. It is useful for grouping values that belong together and should not be structurally changed. Tuples can contain values of different types, and a tuple can contain mutable objects even though the tuple's own references cannot be reassigned.

```python
coordinate = (40.4093, 49.8671)
record = ("station-01", 14, True)
print(coordinate)
print(record)
print(type(coordinate).__name__)
```

## 2. Create tuples correctly

Commas create tuples; parentheses are often used for readability. A one-item tuple needs a trailing comma: `(42,)`. Without the comma, `(42)` is just the integer `42`.

```python
one_item_tuple = (42,)
not_a_tuple = (42)
empty_tuple = ()
print(one_item_tuple, type(one_item_tuple).__name__)
print(not_a_tuple, type(not_a_tuple).__name__)
print(empty_tuple)
```

## 3. Indexing, slicing, and unpacking

Tuples support indexing, slicing, iteration, and unpacking. Unpacking is useful when a function or record returns multiple values.

```python
point = (12, 30, 5)
x, y, z = point
print(point[0])
print(point[1:])
print(x, y, z)

first, *middle, last = (1, 2, 3, 4, 5)
print(first, middle, last)
```

## 4. Tuple methods and operations

Tuples provide `count()` and `index()`. You can also use `len()`, membership checks, concatenation, and repetition. Methods that would mutate the tuple are not available.

```python
values = (2, 4, 2, 6, 2)
print(values.count(2))
print(values.index(6))
print(len(values))
print(2 in values)
print((1, 2) + (3, 4))
```

## 5. When should you use a tuple?

Tuples suit fixed groups such as coordinates, RGB colors, or small records whose positions have a known meaning. For richer records with named fields, a dataclass or another structured type may be clearer. Use a list when the collection itself needs to grow, shrink, or change order.

```python
rgb = (34, 139, 34)
latitude, longitude = (40.4093, 49.8671)
print(f"RGB color: {rgb}")
print(f"Location: {latitude}, {longitude}")
```

## Common mistakes

- `(42)` is not a tuple; `(42,)` is.
- Immutability prevents replacing tuple items, but it does not make nested mutable objects immutable.
- Unpacking requires the number of values to match the target names unless starred unpacking is used.

## 6. Tuples returned from functions

A function can return multiple values by returning a tuple. Callers can store the tuple or unpack its parts into descriptive variables. This is a common pattern for returning a value together with status or metadata.

```python
def min_and_max(values):
    if not values:
        return None, None
    return min(values), max(values)

lowest, highest = min_and_max([8, 3, 11, 5])
print(lowest, highest)
print(min_and_max([]))
```

## 7. Tuples as dictionary keys

Tuples can be dictionary keys or set elements when all their contents are hashable. This makes a tuple useful for representing a fixed coordinate or compound key. A tuple containing a list is not hashable.

```python
cell_counts = {(2, 4): 17, (2, 5): 23}
print(cell_counts[(2, 4)])

try:
    invalid_key = (1, [2, 3])
    hash(invalid_key)
except TypeError:
    print("A tuple containing a list is not hashable.")
```

## 8. Immutability has a boundary

A tuple cannot replace one of its element references, but a mutable object inside it can still change. Also, methods such as `index()` raise `ValueError` when a requested value is missing.

```python
record = ("station-01", [12, 15])
record[1].append(18)
print(record)

values = (2, 4, 6)
try:
    values.index(9)
except ValueError:
    print("The value is not present in the tuple.")
```
