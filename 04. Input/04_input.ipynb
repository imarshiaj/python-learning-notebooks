## 2. Convert user input

Use `int()` for whole numbers and `float()` for decimal values. If the entered text is not a valid number, conversion raises a `ValueError`. Later, the Exception Handling notebook will teach how to handle that gracefully.

```python
distance_text = "12.5"
distance_km = float(distance_text)
count_text = "8"
count = int(count_text)
print(f"Distance: {distance_km} km")
print(f"Count plus one: {count + 1}")
```

## 3. Clean and validate text

`strip()` removes leading and trailing whitespace. Basic checks such as checking for an empty string can prevent common input mistakes. Validation rules should match the real task rather than rejecting harmless input arbitrarily.

```python
raw_name = "  Morgan  "
name = raw_name.strip()
if name:
    print(f"Welcome, {name}!")
else:
    print("A name is required.")
```

## 4. A reusable numeric parsing pattern

A practical program should not assume that users always type valid input. The next cell shows the structure of validation using `try` and `except`; the detailed exception concepts appear in Notebook 17. Notice that the error message helps the user correct the problem.

```python
def parse_positive_integer(text):
    try:
        value = int(text)
    except ValueError:
        return None
    return value if value > 0 else None

for sample in ("12", "0", "three", "-4"):
    parsed = parse_positive_integer(sample)
    print(sample, "->", parsed)
```

## 7. A reliable input workflow

A typical workflow is: read text, remove surrounding whitespace, check whether it is empty, convert it, and validate its allowed range. `input()` always returns a string, even when the user types digits. The examples below use sample strings so the notebook can run automatically.

```python
raw_age = " 21 "
age_text = raw_age.strip()

if not age_text:
    print("Age is required.")
else:
    try:
        age = int(age_text)
    except ValueError:
        print("Age must be a whole number.")
    else:
        if age < 0:
            print("Age cannot be negative.")
        else:
            print("Accepted age:", age)
```

## 8. Convert carefully

`int("3")` succeeds, but `int("3.0")` fails because the string contains a decimal point. If decimal input is allowed, use `float()` and decide how to handle values such as `nan`, infinity, or numbers outside the application's valid range.

```python
print(int("3"))
print(float("3.0"))

for text in ["3.0", "abc", ""]:
    try:
        value = int(text)
    except ValueError:
        print(repr(text), "cannot be converted to int directly")
    else:
        print(value)
```

## 9. A reusable validation pattern

Keep conversion errors separate from business rules. For example, `ValueError` means the text cannot be parsed as a number; a successfully parsed negative number may still be invalid for a particular task.

```python
def parse_positive_quantity(text: str) -> int | None:
    try:
        quantity = int(text.strip())
    except ValueError:
        return None
    return quantity if quantity > 0 else None

for sample in ["5", " 8 ", "0", "-2", "many"]:
    print(repr(sample), "->", parse_positive_quantity(sample))
```
