## 1. Create strings

A string (`str`) is an immutable sequence of Unicode characters. You can use single quotes, double quotes, or triple quotes for multi-line text. Choose quotes that keep the content readable.

```python
single = 'Hello'
double = "Python strings"
multi_line = """A note can
span multiple lines."""
print(single)
print(double)
print(multi_line)
```

## 2. Escape characters and raw strings

An escape sequence begins with a backslash. Common examples include `\n` for a newline, `\t` for a tab, and `\\` for a literal backslash. A raw string such as `r"C:\new\test"` is useful for writing paths and regular-expression patterns because most backslashes are treated literally.

```python
print("First line\nSecond line")
print("Name:\tAvery")
print(r"C:\new\test")
```

## 3. Indexing and slicing

String positions start at zero. Negative indices count backward from the end. A slice `text[start:stop]` includes `start` but excludes `stop`. You may omit either boundary, and a third value can specify a step.

```python
word = "PYTHON"
print(word[0])
print(word[-1])
print(word[1:4])
print(word[:3])
print(word[::2])
print(word[::-1])
```

## 4. String length and membership

Use `len()` to count characters and `in` to test whether a substring appears. These operations are case-sensitive. For case-insensitive matching, normalize the values with a method such as `casefold()` when suitable.

```python
sentence = "Python supports readable code."
print(len(sentence))
print("readable" in sentence)
print("PYTHON" in sentence)
print("PYTHON".casefold() in sentence.casefold())
```

## 5. Useful string methods

Common methods include `lower()`, `upper()`, `title()`, `strip()`, `replace()`, `split()`, `join()`, `startswith()`, and `endswith()`. String methods usually return a new string because strings are immutable.

```python
message = "  Python makes data work easier.  "
clean = message.strip()
print(clean.upper())
print(clean.replace("easier", "more approachable"))
print("alpha,beta,gamma".split(","))
print(" / ".join(["data", "analysis", "python"]))
print(clean.startswith("Python"))
```

## 6. Format strings with f-strings

An f-string starts with `f` and allows expressions inside braces. It is often the most readable way to combine variables and text. Formatting options can control decimal places, alignment, separators, and more.

```python
name = "Riley"
score = 93.4567
population = 1234567
print(f"Hello, {name}!")
print(f"Score: {score:.2f}")
print(f"Population: {population:,}")
print(f"Next year: {2026 + 1}")
```

## 7. Strings are immutable

You cannot change a single character of an existing string in place. Instead, create a new string through slicing, concatenation, or a method. This differs from mutable types such as lists, which are covered next.

```python
language = "python"
capitalized = "P" + language[1:]
print(language)
print(capitalized)
```

## Common mistakes

- The end boundary of a slice is excluded.
- `+` concatenates strings, but it does not automatically convert numbers to text. Use `str(number)` or an f-string.
- `split()` without a separator splits on runs of whitespace; with a separator, adjacent separators can produce empty items.
- String method names are case-sensitive: `strip()` is valid, while `Strip()` is not.

## 9. Practical text cleaning and inspection

Real text often contains extra spaces, inconsistent capitalization, or optional prefixes and suffixes. Use `strip()`, `lower()`/`casefold()`, `startswith()`, `endswith()`, and `count()` where they fit the task.

```python
raw_title = "  URBAN Data Report.PDF  "
title = raw_title.strip()
print(title.lower())
print(title.casefold())
print(title.endswith(".PDF"))
print(title.count("a"))
```

## 10. Split and join text

`split()` turns a string into a list. Without an explicit separator it groups runs of whitespace; with a separator, consecutive separators can create empty strings. `join()` combines an iterable of strings. It does not automatically convert numbers to text.

```python
line = "  park   clinic  school "
print(line.split())
print("red|green||blue".split("|"))
words = ["open", "data", "tools"]
print(" / ".join(words))
print(", ".join(str(number) for number in [3, 5, 8]))
```

## 11. Searching and replacing

`in` checks whether a substring exists. `find()` returns `-1` when no match exists; `index()` instead raises `ValueError`. `replace()` changes all matching occurrences by default and returns a new string.

```python
message = "Python supports text processing. Python is readable."
print("text" in message)
print(message.find("Python"))
print(message.find("Java"))
print(message.replace("Python", "Python 3", 1))
```

## 12. Strings are immutable

String operations create new strings; they do not modify the original string in place. Use slicing or a method and store the result when you need an updated value. Raw strings such as `r"C:\new\reports"` are helpful for Windows paths and regular-expression patterns.

```python
word = "data"
updated = word.upper()
print(word, updated)
print(r"C:\new\reports")
print("Line one\nLine two")
```
