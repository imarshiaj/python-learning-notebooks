## 1. Why use loops?

Loops repeat instructions. A `for` loop iterates over items in an iterable, such as a list, string, tuple, dictionary, or `range`. Use a loop when you need to process several items without duplicating code.

```python
for language in ["Python", "SQL", "R"]:
    print(language)

for character in "code":
    print(character)
```

## 2. Use `range()`

`range(stop)` yields integers starting at zero and stopping before `stop`. You can also provide a `start` and a `step`: `range(start, stop, step)`. The range itself is a lazy sequence-like object, not a list.

```python
print(list(range(5)))
print(list(range(2, 7)))
print(list(range(10, 0, -2)))

for number in range(1, 6):
    print(number, end=" ")
print()
```

## 3. Accumulate values

A loop can update a running total or build a result. For simple numeric aggregation, built-ins such as `sum()` can be clearer, but a loop helps you understand the underlying process.

```python
values = [3, 8, 2, 10, 5]
total = 0
for value in values:
    total += value
print("Total:", total)
print("Average:", total / len(values))
```

## 4. `while` loops

A `while` loop repeats while its condition is true. Make sure something changes so the loop can eventually stop; otherwise, it may run forever. Use `while` when the number of repetitions is not known in advance.

```python
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1
print("Go!")
```

## 5. `break`, `continue`, and `pass`

- `break` exits the innermost loop.
- `continue` skips to the next iteration.
- `pass` is a placeholder that does nothing; it can keep an unfinished block syntactically valid.

```python
for number in range(1, 8):
    if number == 3:
        continue
    if number == 6:
        break
    print(number)
```

## 6. `enumerate()` and `zip()`

`enumerate()` gives each item an index and value. `zip()` combines multiple iterables item by item and stops when the shortest iterable is exhausted.

```python
cities = ["Baku", "Tbilisi", "Yerevan"]
for index, city in enumerate(cities, start=1):
    print(index, city)

projects = ["Atlas", "Beacon"]
status = ["active", "planned"]
for project, state in zip(projects, status):
    print(f"{project}: {state}")
```

## 7. Looping through dictionaries and comprehensions

Loop through a dictionary's `.items()` when both keys and values are needed. Comprehensions are useful for simple transformations and filters, but conventional loops remain valuable for complex logic.

```python
scores = {"Avery": 92, "Morgan": 85, "Riley": 78}
for name, score in scores.items():
    print(f"{name}: {score}")

passing = [name for name, score in scores.items() if score >= 80]
print("Passing:", passing)
```

## Common mistakes

- `range(5)` ends at `4`, not `5`.
- An unchanging `while` condition can create an infinite loop.
- `break` and `continue` affect the innermost loop only.
- If you remove elements from a list while iterating over it, some items may be skipped; consider building a filtered list instead.

## 8. Use `enumerate()` and `zip()`

`enumerate()` provides an index and a value without manually maintaining a counter. `zip()` combines corresponding elements and stops when the shortest iterable is exhausted. In Python 3.10+, `zip(..., strict=True)` raises `ValueError` if lengths differ.

```python
tasks = ["collect", "clean", "analyze"]
for index, task in enumerate(tasks, start=1):
    print(index, task)

names = ["Ava", "Noah"]
scores = [91, 84]
for name, score in zip(names, scores):
    print(name, score)
```

## 9. `break`, `continue`, and loop `else`

- `break` exits the nearest loop.
- `continue` skips the remainder of the current iteration.
- A loop's `else` block runs only when the loop finishes normally, not when it exits via `break`.

The `else` feature is especially useful when searching for an item.

```python
target = 7
for value in [2, 4, 7, 9]:
    if value == target:
        print("Found:", value)
        break
else:
    print("Not found")
```

## 10. Avoid accidental infinite loops and mutation bugs

A `while` loop needs a condition that can eventually become false. When iterating over a list, avoid removing elements from that same list in the loop; build a filtered list instead.

```python
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1

values = [1, 2, 3, 4, 5, 6]
even_values = [value for value in values if value % 2 == 0]
print(even_values)
```
