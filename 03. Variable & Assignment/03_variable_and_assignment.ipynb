## 1. Variables are names bound to objects

A variable is a name that refers to an object. Python is dynamically typed: you do not normally declare a variable's type before assigning a value. The object has a type, and a name can later be rebound to a different object.

```python
city = "Baku"
population_estimate = 2_300_000
city = "Tbilisi"
print(city)
print(population_estimate)
```

## 2. Naming rules and conventions

A valid identifier can contain letters, digits, and underscores, but cannot begin with a digit. Names are case-sensitive and cannot be Python keywords such as `class` or `if`.

Conventions: use `snake_case` for variables and functions; use uppercase names such as `MAX_RETRIES` for constants by convention. Python does not enforce constants simply because a name is uppercase.

```python
user_name = "Sam"
number_of_projects = 3
MAX_RETRIES = 5

print(user_name, number_of_projects, MAX_RETRIES)
```

## 3. Assignment forms

Python supports straightforward assignment, chained assignment, and assigning multiple values at once. Multiple assignment can make related values clear, but avoid packing too many unrelated assignments into one line.

```python
x = 10
y = x
a = b = 0
width, height = 640, 480

print(x, y, a, b)
print(width, height)
```

## 4. Rebinding and augmented assignment

Augmented assignment makes updates compact. It applies an operation and assigns the result back to the target. For mutable objects such as lists, an augmented operation can modify the existing object; this is different from all forms of reassignment in subtle cases.

```python
score = 10
score += 5
score *= 2
print(score)

tags = ["python"]
alias = tags
tags += ["learning"]
print(tags)
print(alias)
```

## 5. Swap values and unpack sequences

Tuple packing and unpacking allow Python to assign multiple values in one statement. A useful idiom is `left, right = right, left`, which swaps two names without a temporary variable.

```python
first, second = "north", "south"
first, second = second, first
print(first, second)

coordinates = (40.4, 49.9)
latitude, longitude = coordinates
print(latitude, longitude)
```

## 6. Inspecting and converting types

Use `type()` to inspect a value. Explicit conversion functions include `int()`, `float()`, `str()`, and `bool()`. Conversion can fail when a text value is not formatted as the requested type; input validation and exception handling will be covered later.

```python
quantity_text = "12"
quantity = int(quantity_text)
price = float("3.50")
label = str(quantity)
print(quantity, type(quantity).__name__)
print(price, type(price).__name__)
print(label, type(label).__name__)
```

## 7. Names, objects, and deletion

Assignment does not necessarily copy an object. When two names refer to the same mutable list, changing the list through either name is visible through both. The `del` statement removes a name binding; it does not guarantee that the object is immediately deleted if other references remain.

```python
original = [1, 2]
reference = original
reference.append(3)
print(original)
del reference
print(original)
```

## Common mistakes

- Using a name before assigning it causes a `NameError`.
- `=` is assignment; `==` compares values.
- A variable name does not permanently lock a value to one type.
- `list_b = list_a` does not make a new list; use `list_a.copy()` or another appropriate copy method when a separate list is required.

## 7. Naming, constants, and type hints

Use descriptive `snake_case` names such as `building_height_m`. Names cannot begin with a digit or contain spaces, and Python keywords such as `class` cannot be used as variable names. Uppercase names are a convention for constants; Python does not enforce immutability. Type hints document expected types but do not automatically validate values.

```python
MAX_RETRIES = 3  # A convention: avoid reassigning this name.
building_height_m: float = 18.5
building_height_m = 20  # Valid Python; type hints are not runtime enforcement.
print(MAX_RETRIES, building_height_m)
```

## 8. Assignment does not always copy an object

Assignment binds a name to an object. If two names refer to the same mutable list, changes through either name are visible through both. Make a shallow copy with `.copy()` when you need a separate outer list.

```python
first = ["north", "south"]
second = first
third = first.copy()
second.append("east")
print("first:", first)
print("third:", third)
```

## 9. Unpacking and swapping values

Sequence unpacking can make assignments concise. The number of targets normally must match the number of values; a mismatch raises `ValueError`. A starred target captures the remaining values in a list.

```python
x, y = 10, 20
x, y = y, x
print("Swapped:", x, y)

first, *middle, last = [1, 2, 3, 4, 5]
print(first, middle, last)
```

### Practical conversion warning

`int()` and `float()` return new numeric values; they do not change the original string. Invalid text raises `ValueError`, so validate or handle it before using the converted value.

```python
raw_count = " 24 "
count = int(raw_count.strip())
print(count + 1)

try:
    int("twenty")
except ValueError:
    print("The text does not represent a valid integer.")
```
