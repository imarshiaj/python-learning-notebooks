## 1. Define and call a function

A function packages reusable behavior behind a name. Define it with `def`, add parentheses for parameters, and end the header with a colon. The indented body runs when the function is called.

```python
def greet(name):
    print(f"Hello, {name}!")

greet("Alex")
greet("Morgan")
```

## 2. Return values

`return` sends a result back to the caller and exits the function. If a function reaches its end without a `return` statement, it returns `None`. Prefer returning values when the caller should decide how to display or use the result.

```python
def rectangle_area(width, height):
    return width * height

area = rectangle_area(4, 3)
print(area)
```

## 3. Parameters and arguments

A function can use positional parameters, keyword arguments, default values, and variable numbers of arguments. Required parameters should generally come before parameters with defaults. Keyword arguments can make calls more readable.

```python
def describe_place(name, city="Baku", *, category="unspecified"):
    return f"{name} is a {category} place in {city}."

print(describe_place("Central Park", category="park"))
print(describe_place("Old Town", "Tbilisi", category="district"))
```

## 4. `*args` and `**kwargs`

Use `*args` to accept extra positional arguments as a tuple. Use `**kwargs` to accept extra keyword arguments as a dictionary. They are useful for flexible APIs, but explicit parameters are often clearer when the interface is known.

```python
def summarize(*numbers):
    return {"count": len(numbers), "total": sum(numbers)}

def show_settings(**settings):
    for key, value in settings.items():
        print(f"{key} = {value}")

print(summarize(2, 4, 6))
show_settings(theme="light", page_size=20)
```

## 5. Docstrings and type hints

A docstring describes what a function does and is usually placed as the first statement in its body. Type hints document expected types and help editors and static analysis tools; Python does not automatically enforce them at runtime.

```python
def estimate_total(unit_cost: float, quantity: int, tax_rate: float = 0.0) -> float:
    """Return the total price including proportional tax."""
    subtotal = unit_cost * quantity
    return subtotal * (1 + tax_rate)

print(estimate_total(12.5, 4, tax_rate=0.1))
print(estimate_total.__doc__)
```

## 6. Scope: local and global names

A name assigned inside a function is normally local to that function. Avoid modifying global state unnecessarily; passing inputs and returning outputs makes functions easier to test and reuse.

```python
tax_rate = 0.1

def total_with_tax(price):
    local_total = price * (1 + tax_rate)
    return local_total

print(total_with_tax(50))
```

## 7. Avoid mutable default arguments

Default argument values are evaluated once when the function is defined, not on every call. A mutable default such as `items=[]` can therefore be shared across calls. Prefer `None` as a sentinel and create a new list inside the function.

```python
def add_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items

print(add_item("a"))
print(add_item("b"))
```

## 8. Small anonymous functions

A `lambda` expression creates a small anonymous function with a single expression. It is sometimes useful as a short sort key. Use `def` for functions that need multiple statements, a docstring, or a descriptive public interface.

```python
places = [("Park", 3), ("Clinic", 1), ("School", 2)]
print(sorted(places, key=lambda place: place[1]))
```

## Common mistakes

- Printing a value inside a function is not the same as returning it.
- A function returns only one object, but that object can be a tuple, list, or dictionary containing multiple values.
- Use descriptive names and keep functions focused.
- Do not use mutable defaults unless sharing the same object across calls is intentional.

## 8. `print()` is not `return`

`print()` displays a value for a person to read. `return` sends a value back to the caller, where it can be saved, combined with another expression, or tested. A function without a `return` statement returns `None`.

```python
def show_total(a, b):
    print(a + b)

def calculate_total(a, b):
    return a + b

shown = show_total(2, 3)
calculated = calculate_total(2, 3)
print("show_total returned:", shown)
print("Reusable result:", calculated * 10)
```

## 9. Avoid mutable default arguments

Default argument values are created when the function is defined, not afresh on each call. A mutable default such as `items=[]` can therefore retain changes between calls. Use `None` as a sentinel and create a new list inside the function.

```python
def add_tag(tag, tags=None):
    if tags is None:
        tags = []
    tags.append(tag)
    return tags

print(add_tag("research"))
print(add_tag("mapping"))
print(add_tag("urban", ["planning"]))
```

## 10. Useful argument patterns

- Positional arguments are matched by position.
- Keyword arguments make calls more readable.
- `*args` collects extra positional arguments into a tuple.
- `**kwargs` collects extra keyword arguments into a dictionary.

Type hints help readers and tools, but Python does not enforce them automatically.

```python
def summarize(title, *values, unit="items", **metadata):
    print(f"{title}: {len(values)} {unit}")
    print("Values:", values)
    print("Metadata:", metadata)

summarize("Measurements", 12, 15, 18, unit="values", city="Baku", verified=True)
```
