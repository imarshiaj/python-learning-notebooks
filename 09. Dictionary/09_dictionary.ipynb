## 1. What is a dictionary?

A dictionary (`dict`) stores key-value pairs. It is useful when values should be retrieved by meaningful keys rather than numeric positions. Keys must be hashable and unique; values can be objects of many types. Dictionaries preserve insertion order in modern Python.

```python
project = {
    "name": "Urban Data Lab",
    "year": 2026,
    "active": True,
}
print(project)
print(project["name"])
```

## 2. Access values safely

Square-bracket lookup raises `KeyError` if a key is missing. The `get()` method returns `None` by default or a default value you provide. Use square brackets when the key is expected to exist; use `get()` when a missing key is a normal possibility.

```python
project = {"name": "Urban Data Lab", "year": 2026}
print(project["name"])
print(project.get("city"))
print(project.get("city", "Not specified"))
```

## 3. Add, update, and delete entries

Assigning to an existing key replaces its value. Assigning to a new key adds it. `update()` merges entries from another mapping or iterable of key-value pairs. Use `pop()` to remove a known key and retrieve its value, or `del` to remove an entry.

```python
profile = {"name": "Taylor", "role": "analyst"}
profile["city"] = "Baku"
profile["role"] = "researcher"
profile.update({"active": True, "level": 1})
removed_role = profile.pop("role")
print(profile)
print("Removed role:", removed_role)
```

## 4. Iterate through dictionaries

Use `.keys()`, `.values()`, and `.items()` to access keys, values, and key-value pairs. Iterating over a dictionary directly iterates over its keys.

```python
metrics = {"accuracy": 0.91, "recall": 0.87, "f1": 0.89}
for name in metrics:
    print(name)
for name, value in metrics.items():
    print(f"{name}: {value:.2f}")
```

## 5. Nested dictionaries

Dictionary values can themselves be dictionaries or lists. This is a common way to represent structured data, including configuration, API responses, and small JSON-like records.

```python
site = {
    "id": "site-001",
    "location": {"city": "Baku", "country": "Azerbaijan"},
    "uses": ["housing", "retail"],
}
print(site["location"]["city"])
print(site["uses"][0])
```

## 6. Dictionary comprehensions

A dictionary comprehension creates key-value pairs from an iterable. Keep the expression easy to read; for complicated transformations, use a loop.

```python
numbers = [1, 2, 3, 4, 5]
squares = {number: number ** 2 for number in numbers}
even_squares = {number: number ** 2 for number in numbers if number % 2 == 0}
print(squares)
print(even_squares)
```

## 7. Count values with a dictionary

A common pattern is to count occurrences. `dict.get(key, 0)` gives a starting count of zero for a key that has not been seen yet. The `collections.Counter` class is another useful option for this task.

```python
land_uses = ["housing", "retail", "housing", "park", "retail", "housing"]
counts = {}
for use in land_uses:
    counts[use] = counts.get(use, 0) + 1
print(counts)
```

## Common mistakes

- Keys must be unique; assigning the same key again updates the existing value.
- A missing key raises `KeyError` when accessed with square brackets.
- Dictionaries are mutable. Be careful when sharing a dictionary between parts of a program.
- Do not mutate a dictionary's size while iterating over it; collect changes separately or iterate over a snapshot when needed.

## 7. Safe lookup: `[]`, `get()`, and membership

`record[key]` raises `KeyError` when a key is absent. `get(key, default)` returns a default instead. The `in` operator checks dictionary keys, not values.

```python
station = {"name": "Central", "capacity": 120}
print(station["name"])
print(station.get("manager", "Not recorded"))
print("capacity" in station)
print(120 in station)  # Checks keys, so this is False.
```

## 8. Update, merge, and count values

`update()` adds or replaces entries. In Python 3.9+, `left | right` creates a merged dictionary; when a key occurs in both, the right-hand value wins. Use `setdefault()` only when you deliberately want to insert a missing key.

```python
defaults = {"theme": "light", "language": "en"}
preferences = {"theme": "dark"}
settings = defaults | preferences
settings.update({"font_size": 14})
settings.setdefault("show_tips", True)
print(settings)
```

## 9. Iterate and transform dictionaries

Use `.items()` for key-value pairs, `.keys()` for keys, and `.values()` for values. Dictionary comprehensions are useful for transforming keys or values. Dictionary keys must be hashable.

```python
scores = {"Ava": 82, "Noah": 91, "Mina": 88}
passed = {name: score for name, score in scores.items() if score >= 85}
print(passed)

try:
    invalid = {[1, 2]: "value"}
except TypeError:
    print("A list cannot be a dictionary key.")
```
