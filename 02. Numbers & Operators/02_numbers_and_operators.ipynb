## 1. Numeric types

Python's main built-in numeric types are:

- `int`: whole numbers, such as `24`, `0`, or `-7`. Integers can grow beyond the range of fixed-width machine integers, subject to available memory.
- `float`: floating-point values, such as `8.36` or `-0.25`. These are approximations for many decimal fractions.
- `complex`: numbers with real and imaginary parts, such as `2 + 6j`. Python uses `j` for the imaginary unit.

Scientific notation uses `e` or `E` to mean “times ten to the power of.” Therefore, `3E5` is `300000.0` and has type `float`.

```python
integer_value = 24
float_value = 8.36
complex_value = 2 + 6j
scientific_value = 3E5

for value in (integer_value, float_value, complex_value, scientific_value):
    print(value, "->", type(value).__name__)
```

## 2. Arithmetic operators

| Operator | Meaning | Example |
|---|---|---|
| `+` | Addition | `2 + 3` |
| `-` | Subtraction | `12 - 8` |
| `*` | Multiplication | `3 * 6` |
| `/` | True division | `24 / 8` |
| `//` | Floor division | `17 // 5` |
| `%` | Remainder | `17 % 5` |
| `**` | Exponentiation | `2 ** 8` |

Use `** 0.5` for a square root of a non-negative number. For exact or advanced numerical work, choose tools appropriate to the problem; floating-point arithmetic has limitations.

```python
print("Addition:", 2 + 3)
print("Subtraction:", 12 - 8)
print("Multiplication:", 3 * 6)
print("True division:", 24 / 8)
print("Floor division:", 17 // 5)
print("Remainder:", 17 % 5)
print("Power:", 2 ** 8)
print("Square root:", 16 ** 0.5)
```

## 3. Understand the three division-related operators

- `/` returns the quotient as a floating-point result.
- `//` performs **floor division**: it rounds the quotient down toward negative infinity, not toward zero.
- `%` returns the remainder associated with floor division. For positive divisors, the remainder is non-negative.

For example, `17 // 5` is `3` and `17 % 5` is `2`. For negative numbers, remember that floor division rounds down: `-17 // 5` is `-4`.

```python
print(17 / 5)
print(17 // 5)
print(17 % 5)
print(-17 // 5)
print(-17 % 5)
print((17 // 5) * 5 + (17 % 5))
```

## 4. Comparison operators

Comparisons produce a Boolean value: `True` or `False`.

- Equal: `==`
- Not equal: `!=`
- Less than: `<`
- Greater than: `>`
- Less than or equal to: `<=`
- Greater than or equal to: `>=`

Do not confuse `=` (assignment) with `==` (comparison).

```python
print(5 == 5)
print(5 != 3)
print(3 < 10)
print(3 <= 3)
print(10 > 4)
print(10 >= 11)
```

## 5. Logical operators

Use `and`, `or`, and `not` to combine or invert conditions. `and` requires both operands to be truthy; `or` requires at least one; `not` reverses truthiness. These operators can short-circuit, so Python may not evaluate every operand.

```python
temperature = 22
is_sunny = True
print(temperature > 15 and is_sunny)
print(temperature < 0 or is_sunny)
print(not is_sunny)
```

## 6. Assignment operators

The assignment operator `=` binds a name to a value. Augmented assignment operators update a variable using an operation: `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, and `**=`. For example, `count += 1` is commonly used to increase a counter.

```python
count = 10
count += 3
print(count)
count *= 2
print(count)
```

## 7. Membership and identity

- `in` and `not in` check membership in a container or sequence.
- `is` and `is not` check object identity: whether two references point to the same object.

Use `==` when you want to compare values. Do not rely on `is` for ordinary numeric or string equality. A common legitimate use of identity is checking for `None` with `value is None`.

```python
print("py" in "python")
print(5 not in [1, 2, 3])

first = [1, 2]
second = [1, 2]
third = first
print(first == second)  # Equal contents
print(first is second)  # Different list objects
print(first is third)   # Same list object
value = None
print(value is None)
```

## 8. Bitwise operators

Bitwise operators act on integer bits. They are useful in low-level work, masks, and some performance-sensitive or protocol-related tasks. For example, `5` is `0b0101` and `3` is `0b0011`.

- `&`: bitwise AND
- `|`: bitwise OR
- `^`: bitwise XOR
- `~`: bitwise inversion
- `<<`: shift bits left
- `>>`: shift bits right

Note carefully: `<<` is left shift and `>>` is right shift.

```python
print(5 & 3)
print(5 | 3)
print(5 ^ 3)
print(5 << 1)
print(5 >> 1)
print(bin(5), bin(3))
```

## 9. Operator precedence

Python follows a precedence order. Parentheses make the intended grouping explicit and improve readability. A simplified order, from higher to lower precedence, is: parentheses/grouping; exponentiation; unary `+`, `-`, `~`; multiplication, division, floor division, remainder; addition and subtraction; shifts; bitwise operators; comparisons and membership/identity tests; `not`; `and`; `or`; assignment. The full language reference has the complete table.

Exponentiation groups from right to left: `2 ** 3 ** 2` means `2 ** (3 ** 2)`.

```python
print(4 + 6 * 5 - 3)
print((4 + 6) * 5 - 3)
print(4 + 6 * (5 - 3))
print((4 + 6) * (5 - 3))
print(2 ** 3 ** 2)
```

## 10. Floating-point precision

Many decimal fractions cannot be represented exactly in binary floating point. This is why `0.1 + 0.2` may display as `0.30000000000000004`. For comparisons involving approximate measurements, use a tolerance such as `math.isclose()`. For decimal arithmetic that must follow decimal rules (for example, some currency calculations), consider `decimal.Decimal` and construct it from strings.

```python
import math
from decimal import Decimal

print(0.1 + 0.2)
print(math.isclose(0.1 + 0.2, 0.3))
print(Decimal("0.1") + Decimal("0.2"))
```

## 10. Practical numeric edge cases

### Floor division and modulo with negative numbers

`//` rounds the quotient down toward negative infinity; it does not simply remove the decimal part. Python's modulo result is consistent with the divisor's sign.

```python
print(7 // 3)       # 2
print(-7 // 3)      # -3, not -2
print(7 % 3)        # 1
print(-7 % 3)       # 2
print(divmod(17, 5))  # (quotient, remainder)
```

### Floating-point precision

Most decimal fractions cannot be represented exactly as binary floating-point numbers. For approximate comparisons, `math.isclose()` is often more reliable than `==`. For simple currency calculations, `Decimal` constructed from strings can be useful.

```python
import math
from decimal import Decimal

print(0.1 + 0.2 == 0.3)
print(math.isclose(0.1 + 0.2, 0.3))
print(Decimal("0.10") + Decimal("0.20"))
```

### Rounding, formatting, and conversion are different

- `round(value, digits)` returns a rounded numeric value, using Python's rounding rules.
- An f-string format such as `:.2f` controls how a number is displayed.
- `int("12")` works, but `int("12.5")` raises `ValueError`; convert through `float` when decimal input is intended.

```python
price = 12.345
print(round(price, 2))
print(f"Displayed price: {price:.2f}")
print(int(float("12.5")))
```

### Chained comparisons and operator meaning

Python supports mathematical-looking chains such as `0 <= value < 100`. Use `==` to compare values and `is` to test object identity. The bitwise operator `^` is XOR, not exponentiation; use `**` for powers.

```python
value = 42
print(0 <= value < 100)
print(2 ** 5)          # power: 32
print(5 ^ 3)           # bitwise XOR: 6
print(True == 1)       # True, because bool is a numeric subtype
print(True is 1)       # False: not the same object
```
