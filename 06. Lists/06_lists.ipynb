## 1. What is a list?

A list is an ordered, mutable collection. It can contain values of different types, although lists are often easier to work with when their elements have a consistent purpose. Lists are written with square brackets.

```python
languages = ["Python", "SQL", "JavaScript"]
measurements = [12, 15, 13, 18]
mixed = ["site A", 4, True]
print(languages)
print(measurements)
print(mixed)
```

## 2. Indexing and slicing

Lists support zero-based indexing, negative indexes, and slicing just like strings. An index outside the valid range raises `IndexError`.

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[0])
print(numbers[-1])
print(numbers[1:4])
print(numbers[:3])
print(numbers[::2])
```

## 3. Modify a list

Lists are mutable. You can replace an item, append one item, extend with several items, insert at an index, or remove elements. `append()` adds one object; `extend()` adds each item from an iterable.

```python
tasks = ["read", "practice"]
tasks.append("review")
tasks.extend(["plan", "build"])
tasks.insert(1, "take notes")
tasks[0] = "read documentation"
print(tasks)
```

## 4. Remove and search for elements

`remove(value)` removes the first matching value and raises `ValueError` if it is absent. `pop(index)` removes and returns an item (the last item by default). `del` can remove an item or slice. Use `in` to test membership.

```python
colors = ["red", "blue", "green", "blue"]
colors.remove("blue")
last_color = colors.pop()
print(colors)
print("Removed:", last_color)
print("red" in colors)
```

## 5. Useful list operations

Use `len()`, `min()`, `max()`, and `sum()` where appropriate. Lists of comparable values can be sorted in place with `sort()` or copied into sorted order using `sorted()`. `reverse()` changes a list in place.

```python
scores = [78, 92, 85, 92, 64]
print("Count:", len(scores))
print("Total:", sum(scores))
print("Highest:", max(scores))
print("Sorted copy:", sorted(scores))
scores.sort(reverse=True)
print("Original after sort:", scores)
```

## 6. Copying lists and aliasing

`second = first` creates another name for the same list. Use `first.copy()` or `first[:]` for a shallow copy. A shallow copy creates a new outer list, but nested mutable objects are still shared.

```python
original = [1, 2, 3]
alias = original
copy_of_original = original.copy()
alias.append(4)
print("Original:", original)
print("Alias:", alias)
print("Copy:", copy_of_original)
```

## 7. List comprehensions

A list comprehension creates a new list from an iterable, optionally filtering items. It is concise, but a normal `for` loop may be clearer when the transformation is complicated.

```python
squares = [number ** 2 for number in range(1, 6)]
even_squares = [number ** 2 for number in range(1, 11) if number % 2 == 0]
print(squares)
print(even_squares)
```

## 8. Lists of lists

Lists can contain other lists. This can represent rows and columns, but pay attention to aliasing when creating nested lists. A safe common pattern for a grid of independent rows is a comprehension such as `[[0] * columns for _ in range(rows)]`, not `[[0] * columns] * rows`.

```python
grid = [[0] * 3 for _ in range(2)]
grid[0][1] = 9
print(grid)
```

## Common mistakes

- Indexing starts at zero.
- `sort()` mutates the list and returns `None`; use `sorted()` when you need a sorted copy.
- Do not modify a list while iterating over it unless you understand the consequences; building a new list is often safer.
- A shallow copy does not recursively copy nested objects.

## 9. Common list methods: what do they return?

`append()`, `extend()`, `sort()`, and `reverse()` modify a list in place and return `None`. `sorted()` creates a new sorted list. This distinction matters when assigning a method's result.

```python
scores = [82, 67, 91]
returned_value = scores.sort()
print(scores)
print(returned_value)  # None
print(sorted(scores, reverse=True))
```

## 10. Sort records with a key

The `key` argument lets you sort by a chosen property, such as a name, date, or numeric measurement. For text, `.lower()` can provide case-insensitive sorting.

```python
places = [
    {"name": "Central Park", "area": 12.5},
    {"name": "Riverside Park", "area": 8.2},
    {"name": "Old Town Square", "area": 1.4},
]
for place in sorted(places, key=lambda item: item["area"], reverse=True):
    print(place["name"], place["area"])
```

## 11. Safe removal and filtering

`pop()` on an empty list and `remove()` for a missing value raise `IndexError` and `ValueError`, respectively. Check the list or use a `try` block when absence is expected. When filtering, a list comprehension often avoids the pitfalls of removing items while iterating.

```python
items = ["map", "survey", "map", "report"]
items = [item for item in items if item != "map"]
print(items)

values = []
print(values.pop() if values else "Nothing to remove")
```

## 12. `enumerate()` and nested-list independence

Use `enumerate()` when you need both an index and an element. When creating a grid, avoid multiplying a nested list (`[[0] * 3] * 2`) because both rows refer to the same inner list.

```python
for index, item in enumerate(["survey", "clean", "analyze"], start=1):
    print(index, item)

safe_grid = [[0] * 3 for _ in range(2)]
safe_grid[0][0] = 7
print(safe_grid)
```
