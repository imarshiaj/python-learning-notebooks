## 1. What is a set?

A set is an unordered collection of distinct, hashable elements. Duplicate values are removed automatically. Because a set is unordered, do not rely on its display order. Use a list or tuple if sequence order matters.

```python
service_types = {"park", "clinic", "school", "park", "library"}
print(service_types)
print("Unique count:", len(service_types))
```

## 2. Create sets

Use curly braces for a non-empty set or `set(iterable)` to build one from an iterable. Empty braces `{}` create an empty dictionary, not an empty set; use `set()` for an empty set.

```python
empty_set = set()
letters = set("banana")
from_list = set([1, 2, 2, 3, 3, 3])
print(empty_set)
print(letters)
print(from_list)
```

## 3. Add and remove elements

Use `add()` to insert one item and `update()` to insert items from an iterable. `discard()` does nothing if an element is absent; `remove()` raises `KeyError` when the element is missing. `pop()` removes and returns an arbitrary element.

```python
categories = {"housing", "retail"}
categories.add("public space")
categories.update(["transport", "healthcare"])
categories.discard("industrial")
print(categories)
```

## 4. Set operations

Set operations help compare groups: union (`|`), intersection (`&`), difference (`-`), and symmetric difference (`^`). Method forms such as `.union()` and `.intersection()` are also available.

```python
planned = {"park", "clinic", "school"}
existing = {"clinic", "library"}
print("Union:", planned | existing)
print("Intersection:", planned & existing)
print("Difference:", planned - existing)
print("Symmetric difference:", planned ^ existing)
```

## 5. Subsets, supersets, and membership

Use `issubset()`, `issuperset()`, or the operators `<=` and `>=` to compare sets by inclusion. Use `in` to test membership.

```python
required = {"access", "safety"}
project_features = {"access", "safety", "green space"}
print(required <= project_features)
print(project_features.issuperset(required))
print("safety" in project_features)
```

## 6. Set comprehensions and frozen sets

Set comprehensions use curly braces with an expression and optional filtering. A `frozenset` is an immutable set; it can itself be used as an element of another set or as a dictionary key because it is hashable.

```python
squares = {number ** 2 for number in range(1, 6)}
fixed_codes = frozenset({"A", "B", "C"})
print(squares)
print(fixed_codes)
```

## Common mistakes

- Set display order is not guaranteed.
- Set elements must be hashable: integers and strings work, but ordinary lists and dictionaries cannot be elements.
- `{}` creates a dictionary; `set()` creates an empty set.
- Use a set when uniqueness and membership testing are important, not when you need to preserve duplicates or positions.

## 7. Remove safely and handle missing items

`remove(value)` raises `KeyError` if the value does not exist. `discard(value)` is useful when absence is expected because it does nothing in that case. A set's displayed order is not guaranteed.

```python
permissions = {"read", "write"}
permissions.discard("admin")
print(permissions)

try:
    permissions.remove("admin")
except KeyError:
    print("The requested permission was not present.")
```

## 8. Set elements must be hashable

Immutable values such as strings, integers, and suitable tuples can be set elements. Lists and dictionaries are mutable and unhashable, so trying to add one raises `TypeError`. Use a tuple instead of a list when the contents are fixed and hashable.

```python
coordinates = {(10, 20), (30, 40)}
print((10, 20) in coordinates)

try:
    invalid = {[1, 2], [3, 4]}
except TypeError:
    print("Lists cannot be elements of a set.")
```

## 9. Remove duplicates while preserving order

A set is excellent for uniqueness when order does not matter. When the original order should remain, use `dict.fromkeys()` for a simple sequence of hashable values.

```python
categories = ["park", "clinic", "park", "school", "clinic"]
unique_in_order = list(dict.fromkeys(categories))
print(unique_in_order)
```
