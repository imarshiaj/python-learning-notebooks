## 1. Classes and objects

Object-oriented programming (OOP) groups data and behavior into objects. A **class** defines a kind of object; an **instance** is a specific object created from that class. Use classes when they help model a meaningful entity or keep related state and operations together.

```python
class Dog:
    def bark(self):
        return "Woof!"

pet = Dog()
print(pet.bark())
```

## 2. Initialize instance attributes

The `__init__()` method initializes a new instance. The first parameter, conventionally named `self`, refers to the instance. Instance attributes belong to each object and are commonly created in `__init__()`.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"My name is {self.name} and I am {self.age}."

person = Person("Avery", 29)
print(person.introduce())
```

## 3. Methods and object state

Methods are functions defined in a class. An instance method uses `self` to access or update the instance's state. Keep an object's invariants consistent: if an attribute must always meet a rule, validate changes at the object's boundary.

```python
class Counter:
    def __init__(self, start=0):
        self.value = start

    def increment(self, amount=1):
        self.value += amount
        return self.value

counter = Counter(10)
print(counter.increment())
print(counter.increment(5))
```

## 4. Class attributes and instance attributes

A class attribute is stored on the class and shared unless shadowed by an instance attribute. An instance attribute belongs to one object. Do not use a mutable class attribute for data that should be independent for every instance.

```python
class Project:
    category = "research"

    def __init__(self, name):
        self.name = name

first = Project("Atlas")
second = Project("Beacon")
print(first.category, second.category)
print(first.name, second.name)
```

## 5. Encapsulation and properties

Python conventionally uses a leading underscore, such as `_balance`, to indicate an implementation detail. A `property` can expose a method through attribute-style access and validate updates. Python does not use strict private access in the same way as some other languages.

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature cannot be below absolute zero.")
        self._celsius = value

temperature = Temperature(20)
temperature.celsius = 25
print(temperature.celsius)
```

## 6. Inheritance and overriding

Inheritance lets a class extend another class. A subclass can override a method to specialize behavior. Use `super()` when you want to call a parent implementation. Prefer composition over deep inheritance when an object can simply contain or use another object.

```python
class Animal:
    def speak(self):
        return "Some sound"

class Cat(Animal):
    def speak(self):
        return "Meow"

class Dog(Animal):
    def speak(self):
        return "Woof"

for animal in (Cat(), Dog()):
    print(animal.speak())
```

## 7. Class methods and static methods

A `@classmethod` receives the class as `cls` and is often used for alternative constructors. A `@staticmethod` receives neither `self` nor `cls` automatically and is used for a related utility that does not need instance or class state. Use ordinary module-level functions when a method does not conceptually belong to the class.

```python
class ProjectCode:
    prefix = "PRJ"

    def __init__(self, identifier):
        self.identifier = identifier

    @classmethod
    def from_number(cls, number):
        return cls(f"{cls.prefix}-{number:04d}")

    @staticmethod
    def is_valid(text):
        return text.startswith("PRJ-")

code = ProjectCode.from_number(27)
print(code.identifier)
print(ProjectCode.is_valid(code.identifier))
```

## 8. Dataclasses for data-focused objects

When a class mainly stores data, `dataclasses.dataclass` can generate methods such as `__init__()` and a helpful representation. Dataclasses work well for records and configuration objects. Add methods when they represent behavior that belongs naturally to the object.

```python
from dataclasses import dataclass

@dataclass
class Rectangle:
    width: float
    height: float

    def area(self) -> float:
        return self.width * self.height

    def perimeter(self) -> float:
        return 2 * (self.width + self.height)

shape = Rectangle(4, 3)
print(shape)
print("Area:", shape.area())
print("Perimeter:", shape.perimeter())
```

## 9. Composition

Composition models a “has-a” relationship: one object contains or works with another. It can keep designs flexible and reduce the need for complicated inheritance trees.

```python
class Engine:
    def start(self):
        return "Engine started"

class Vehicle:
    def __init__(self, engine):
        self.engine = engine

    def start(self):
        return self.engine.start()

vehicle = Vehicle(Engine())
print(vehicle.start())
```

## Common mistakes and design tips

- Always include `self` as the first parameter of ordinary instance methods.
- Do not put per-instance mutable state in a class attribute.
- Use classes when they clarify a model; not every small script needs OOP.
- Prefer descriptive names, small methods, and clear responsibilities.
- Use dataclasses for straightforward data records, and consider composition before building deep inheritance hierarchies.

## 8. Instance attributes and class attributes

Instance attributes belong to one object; class attributes are shared through the class unless shadowed by an instance attribute. Be careful with mutable class attributes such as lists, because all instances may modify the same list.

```python
class Sensor:
    unit = "°C"  # Class attribute shared by instances.

    def __init__(self, name, value):
        self.name = name
        self.value = value

first = Sensor("north", 18.2)
second = Sensor("south", 21.4)
print(first.unit, second.unit)
first.unit = "F"
print(first.unit, second.unit)
```

## 9. Encapsulation with properties

A property allows attribute-style access while keeping validation in one place. It can prevent invalid state from entering an object. Avoid adding properties or classes when a simple variable or function would be clearer.

```python
class Temperature:
    def __init__(self, celsius):
        self.celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature is below absolute zero.")
        self._celsius = value

reading = Temperature(20)
reading.celsius = 22
print(reading.celsius)
```

## 10. Class methods, static methods, and inheritance

- An instance method receives `self` and works with one object's state.
- A class method receives `cls` and is often used as an alternate constructor.
- A static method receives neither automatically and is useful for a helper logically grouped with the class.
- Inheritance models an “is-a” relationship; composition models a “has-a” relationship.

```python
class Project:
    category = "research"

    def __init__(self, name):
        self.name = name

    @classmethod
    def from_slug(cls, slug):
        return cls(slug.replace("-", " ").title())

    @staticmethod
    def normalize_name(name):
        return " ".join(name.split()).title()

project = Project.from_slug("urban-data-lab")
print(project.name)
print(Project.normalize_name("  city   analytics "))
```

## 11. Dataclasses and object design

`@dataclass` can generate an initializer and readable representation for classes that mainly store data. Use `frozen=True` when attribute reassignment should be blocked. It does not recursively freeze nested mutable objects. Prefer composition over complex inheritance trees when objects have distinct responsibilities.

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class Coordinate:
    x: float
    y: float

point = Coordinate(10.5, 20.0)
print(point)
print(isinstance(point, Coordinate))
```

### Common object-oriented mistakes

- Include `self` as the first parameter of ordinary instance methods.
- Use `self.attribute` for per-instance state; assigning to a bare local name does not create an instance attribute.
- Validate data at appropriate boundaries, but avoid adding excessive defensive code for impossible states.
- Prefer clear, small classes with one responsibility over a large class that manages unrelated tasks.
