## 1. Iterable, iterator, and generator

An **iterable** is an object you can loop over, such as a list or string. An **iterator** produces items one at a time. A **generator** is a convenient way to create an iterator, often with a function containing `yield` or a generator expression.

## 2. `yield` pauses and resumes a function

A generator function does not run all its body when called. It returns a generator object. Each request for the next item runs until the next `yield`, which provides a value and pauses the function's state.

```python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1

sequence = count_up_to(4)
print(next(sequence))
print(next(sequence))
print(list(sequence))
```

## 3. Generate a stream of results

Generators are useful when a sequence is large, infinite, or expensive to compute. They can avoid building a full list in memory. A generator normally produces each item only when requested.

```python
def running_totals(numbers):
    total = 0
    for number in numbers:
        total += number
        yield total

print(list(running_totals([2, 4, 6, 8])))
```

## 4. Generator expressions

A generator expression resembles a list comprehension but uses parentheses. It generates items lazily rather than creating the entire list immediately. Choose a list comprehension when you need to reuse or inspect all results; choose a generator when streaming is useful.

```python
squares_list = [number ** 2 for number in range(1, 6)]
squares_generator = (number ** 2 for number in range(1, 6))
print(squares_list)
print(next(squares_generator))
print(list(squares_generator))
```

## 5. Generators are single-pass

Once a generator is exhausted, iterating over it again produces no values. If you need to repeat the computation, call the generator function again or store the results in a list when that is appropriate.

```python
def letters():
    yield "A"
    yield "B"
    yield "C"

items = letters()
print(list(items))
print(list(items))
print(list(letters()))
```

## 6. Compose lazy processing steps

Generator expressions can form a simple processing pipeline. The next cell filters and transforms numbers without creating intermediate lists. This is a small example of lazy evaluation.

```python
values = range(1, 11)
evens = (value for value in values if value % 2 == 0)
squares = (value ** 2 for value in evens)
print(list(squares))
```

## 7. When not to use a generator

A generator is not automatically better than a list. Lists support indexing and repeated iteration, while generators are typically single-pass and do not provide list-style random access. Choose the simplest tool that matches the amount and usage pattern of the data.

## Common mistakes

- Calling a generator function returns a generator object; it does not immediately return the final list.
- A generator is consumed as it is iterated over.
- `yield` produces one value at a time; `return` ends the generator.
- A generator can only save memory when the processing does not also retain all generated values elsewhere.

## 6. Iterables, iterators, and generators

An iterable can produce an iterator; an iterator returns one item at a time with `next()`. Lists are reusable collections, while generators are usually single-pass iterators that compute values as they are requested.

```python
numbers = [10, 20, 30]
iterator = iter(numbers)
print(next(iterator))
print(next(iterator))
print(list(numbers))  # The list can be iterated again.
```

## 7. Generator exhaustion and `next()`

A generator raises `StopIteration` when exhausted. `next(generator, default)` lets you provide a fallback instead of receiving that exception. Calling `list()` consumes all remaining generator values.

```python
def short_sequence():
    yield "first"
    yield "second"

items = short_sequence()
print(next(items, "finished"))
print(next(items, "finished"))
print(next(items, "finished"))
```

## 8. Lazy processing and generator expressions

A generator expression uses parentheses and does not build the entire result list immediately. Lazy processing can reduce memory use for large inputs, but it also means values are only computed when requested.

```python
squares = (number * number for number in range(1, 6))
print(next(squares))
print(sum(squares))  # Consumes the remaining values.
```

## 9. Delegate iteration with `yield from`

`yield from` forwards values from another iterable or generator. It is useful for composing small generators. Keep in mind that a generator cannot be indexed or measured with `len()` unless you first materialize its values.

```python
def combined():
    yield from ["roads", "buildings"]
    yield from ("services", "parks")

print(list(combined()))
```
